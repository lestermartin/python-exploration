## Control Flow: `if` / `elif` / `else`

* Branch on a condidtion to class to classify or react to data
* elif chains are preferred over separate if statements when branches are mutually exclusive
* Always consider the else -- what happens when nothing matches?


```python
# run this with diff vals for 'side'
side = "BUY"

# simple if/then/else
if side == "BUY":
  print("Buying")
elif side == "SELL":
  print("Selling")
else:
  print("Unknown side:", side)
```

## Control Flow: `for` Loops

* `for` loops iterate over a known collection
* The workhorse of data processing in Python
* Every trade in the list is visited exactly once, in order

```python
# more on this data structure later
trades = [
    {"trade_id": "T0001", "side": "BUY", "qty": 120, "price": 185.32},
    {"trade_id": "T0002", "side": "BUY", "qty": 60, "price": 402.11},
]

total = 0.0
for trade in trades:
  value = trade["qty"] * trade["price"]
  total += value

print(f"Total: {total:,.2f}")
```

## Control Flow: `while` Loops

* `while` loops repeat until a condition becomes false
* Use far less often than `for` loops in data work
* Easy to get wrong: forgetting to update `i` causes and infinite loop

```python
# more on lists later
trades = [1, 2, 3, 4, 5]
i = 0
count = 0

while i < len(trades):
  count += 1
  i += 1

print(f"Processed {count} trades")
```

## Writing Functions, with Type Hints

* A function gives a name to logic you'd otherwise repeat
* `quantity: float` and `-> float` are type hints: not enforced at runtime, but self-documenting
* `threshold=20000` is a default argument — callers can omit it

```python
# a method
def trade_value(
    quantity: float, price: float
) -> float:
    return quantity * price

# another method
def classify_trade(
    value: float,
    threshold: float = 20000
) -> str:
    if value > threshold:
        return "large"
    return "normal"
```

## Calling Functions: Positional & Keyword Arguments

* Positional arguments are matched by position
* Keyword arguments are matched by name, and can override a default
* Prefer keyword arguments when a call's meaning isn't obvious from position alone

```python
print(" ... A) trade_value with positional argument")
value = trade_value(120, 185.32)
print(value)
print()

print(" ... B) classify_trade with default threshold of 20000")
print(classify_trade(value))
print()

print(" ... C) classify_trade with function as an argument and positional 2nd argument")
print(classify_trade(trade_value(120, 185.32), 50000))
print()

print(" ... D) classify_trade with keyword argument")
print(classify_trade(22238.40, threshold=50000))
print()
```

### Raises Custom Exceptions

* A custom exception is a class that inherits from `Exception` and describes a business-rule violation in your domain's own language
* Built-in exceptions (`TypeError`) describe a Python-level mistake

```python
class InvalidTradeError(Exception):
    """A trade failed validation."""

def trade_value(quantity, price):
    if quantity <= 0:
        raise InvalidTradeError(
            f"bad quantity: {quantity}")
    return quantity * price
```

```python
# raises built-in Python exception (TypeError)
value = trade_value("120", 185.32)
```

```python
# raises the custom exception (InvalidTradeError)
value = trade_value(-120, 185.32)
```

### Multiple `except` Blocks

* Different failure kinds deserve different handling
* One `except` block can catch multiple types of exceptions
* Be mindful of ordering as once an exception is caught, the other blocks are skipped

```python
# catches the built-in exception
try:
    value = trade_value(-120, 185.32)
except (TypeError, ValueError) as e:
    print(f"Bad data type: {e}")
except InvalidTradeError as e:
    print(f"Rule violation: {e}")
```

```python
# catches the custom exception
try:
    value = trade_value("120", 185.32)
except (TypeError, ValueError) as e:
    print(f"Bad data type: {e}")
except InvalidTradeError as e:
    print(f"Rule violation: {e}")
```

### `else` and `finally`

* `else` runs only when the `try` block succeeds -- keep success logic visibly separate from failure logic
* `finally` always runs, success or failure -- the right place for unconditional cleanup
* A common bug is to put success-only logic in the `finally` by mistake

```python
# happy path
try:
    value = trade_value(120, 185.32)
except InvalidTradeError as e:
    print(f"Skipped: {e}")
else:
    total += value  # only on success
    print(f"Total = {total}")
finally:
    print("** ALWAYS executed")
```

```python
# exception path
try:
    value = trade_value(-120, 185.32)
except InvalidTradeError as e:
    print(f"Skipped: {e}")
else:
    total += value  # only on success
    print(f"Total = {total}")
finally:
    print("** ALWAYS executed")
```

### Re-raising with Context

* Sometimes the right response isn't to swallow an exception, it's to add context and keep propagating
* `raise ... from e` preserves the original exception as the visible cause
* Here: the `trade_id` is added, since the inner function didn't know it

```python
# note: more on dictionaries later
def safe_trade_value(trade: dict) -> float:
    try:
        return trade_value(
            trade["quantity"], trade["price"])
    except InvalidTradeError as e:
        raise InvalidTradeError(
            f"{trade['trade_id']}: {e}"
        ) from e
```

```python
# here is a simple dictionary (of KVPs)
trade_one = {
    "trade_id": "T0001",
    "quantity": 120,
    "price": 185.32,
}

safe_trade_value(trade_one)
```

```python
# another dict but invalid data this time
trade_two = {
    "trade_id": "T0002",
    "quantity": -120,
    "price": 185.32,
}

safe_trade_value(trade_two)
```

# Pandas examples

This notebook is focused on the Pandas examples in Sprint 4.


```python
import pandas as pd
import urllib.request
url = "https://raw.githubusercontent.com/Neueda-Technologies/leap-sprint4/refs/heads/main/shared/trades.csv"
urllib.request.urlretrieve(url, "trades.csv")
```

```python
!cat trades.csv
```

```python
trades = pd.read_csv("trades.csv")
trades
```

```python
trades["value"].skew()
```

```python
pd.cut(trades["value"], bins=5).value_counts().sort_index()
```

```python
trades.groupby("advisor")["value"].agg(
    ["count", "sum", "mean"])
```

```python
trades.groupby("advisor").agg({
    "value": ["count", "sum", "mean"],
    "quantity": "sum",
    "price": ["median","max"],
    "instrument": "count",
})
```

```python
pd.crosstab(trades["asset_class"], trades["side"])
```

```python
trades.pivot_table(
    index="asset_class", columns="side",
    values="value", aggfunc="count",
    fill_value=0,
)
```

```python
trades.pivot_table(
    index="asset_class", columns=["side","client_id"],
    values="value", aggfunc="count",
    fill_value=0,
)
```

```python
trades.pivot_table(
    index="asset_class", columns=["side","client_id", "instrument"],
    values="value", aggfunc=["sum","mean"],
    fill_value=0,
)
```

```python
trades.pivot_table(
    index="asset_class", columns=["side","client_id"],
    values="value", aggfunc=["count","sum"],
    fill_value=0,
)
```

```python
trades.head()
```

```python
trades2 = pd.read_csv("trades.csv", parse_dates=["trade_date"])
by_date = trades2.set_index("trade_date")
by_date.head()
```

```python
by_week = by_date["value"].resample("D").sum()
by_week
```

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

totals = trades.groupby("asset_class")["value"].sum().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(totals.index, totals.values, color="#1E6FA8")
ax.set_title("Total Trade Value by Asset Class ($)")
ax.set_xlabel("Asset Class")
ax.set_ylabel("Total Value ($)")
ax.set_ylim(bottom=0)
fig.tight_layout()
fig
```

```python
fig, ax = plt.subplots()
ax.plot(by_week.index, by_week.values,
        marker="o")
fig.autofmt_xdate()
ax.set_title("Total Trade Value by Week")
ax.set_xlabel("Week")
ax.set_ylabel("Total Value ($)")
fig.tight_layout()
fig
```

```python


df = pd.DataFrame({
    "department": ["Sales", "Sales", "Sales", "Eng", "Eng", "Eng", "HR"],
    "employee":   ["Amy", "Bob", "Cara", "Dan", "Eve", "Finn", "Gia"],
    "salary":     [70000, 65000, 72000, 95000, 105000, 98000, 60000],
    "bonus":      [5000, 4000, 6000, 8000, 10000, 7000, 3000],
    "age":        [29, 41, 35, 27, 33, 45, 38],
    "state":      ["TX", "GA", "TX", "GA", "TX", "GA", "TX"],
})
```

```python
single_col_aggs = df.groupby("department")["salary"].agg(
    ["mean", "min", "max"]
)
print(single_col_aggs)
```

```python
single_col_aggs = df.groupby("department")["salary"].agg(
    ["mean", "min", "max"]
)
print(single_col_aggs)
```

```python
multi_cols_aggs = df.groupby("department").agg({
    "salary": ["mean", "min", "max"],
    "bonus": "sum",
    "age": ["median","max"],
    "employee": "count",
})
multi_cols_aggs
```

```python
pd.crosstab(df["department"], df["state"])
```

```python
df.pivot_table(
    index="department", columns="state",
    values="salary", aggfunc="sum",
    fill_value=0,
)
```

```python
df.pivot_table(
    index="department", columns="state",
    values="salary", aggfunc="sum",
    fill_value=0,
)
```

```python
df.pivot_table(
    index="department", columns=["state","age"],
    values="salary", aggfunc="sum",
    fill_value=0,
)
```

```python
df.pivot_table(
    index="department", columns=["state","age","bonus"],
    values="salary", aggfunc="sum",
    fill_value=0,
)
```

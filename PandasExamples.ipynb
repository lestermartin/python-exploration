# Pandas examples

This notebook is focused on the Pandas examples in Sprint 4.


```python
import pandas as pd

df = pd.DataFrame({
    "department": ["Sales", "Sales", "Sales", "Eng", "Eng", "Eng", "HR"],
    "employee":   ["Amy", "Bob", "Cara", "Dan", "Eve", "Finn", "Gia"],
    "salary":     [70000, 65000, 72000, 95000, 105000, 98000, 60000],
    "bonus":      [5000, 4000, 6000, 8000, 10000, 7000, 3000],
    "age":        [29, 41, 35, 27, 33, 45, 38],
})

summary = df.groupby("department").agg({
    "salary": ["mean", "min", "max"],
    "bonus": "sum",
    "age": ["median","max"],
    "employee": "count",
})
print(summary)
```
